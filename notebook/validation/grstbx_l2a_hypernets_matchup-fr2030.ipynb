# FR2030 L2A products: remote-sensing reflectance against HYPERNETS

This notebook evaluates the remote-sensing reflectance $R_{rs}$ of the Sentinel-2 L2A products of the
**FR2030** processing chain (GRS run in production, products `S2_L2A_..._v01_00_01`, Zarr format)
against the measurements of the [HYPERNETS](https://www.hypernets.eu) autonomous hyperspectral
radiometer installed in the Berre lagoon (France), site `BEFR`, and compares them with the GRS products
used in the tutorial *L2A validation: remote-sensing reflectance against HYPERNETS*.

**Workflow**

1. load the HYPERNETS time series and normalize the in situ $R_{rs}$ to nadir viewing,
2. build the multi-temporal datacube of the FR2030 products over a box centred on the station
   ({class}`grstbx.L2grs`, which opens the Zarr stores directly) and compute, for each image, the
   statistics of the valid pixels within the box,
3. pair each image with the in situ spectra acquired within a given time lag and compute the
   agreement statistics,
4. compare the FR2030 and GRS reflectances of the common dates.

```{note}
The FR2030 products and the HYPERNETS data are not distributed with grstbx: the notebook is rendered
with the outputs of its last run. Set the paths of the configuration cell to run it on your own data.
```

## HYPERNETS data

The individual HYPERNETS files (`HYPERNETS_DIR/SITE/YYYY/MM/DD/<sequence>/*.nc`) are concatenated
once into `HYPERNETS_DIR/SITE/SITE_hypernets.nc`.

````python
def build_hypernets_file(site, ofile, chunk=50):
    """Concatenate the HYPERNETS L2A files of a site along time into a single netCDF file."""
    files = sorted(glob.glob(opj(HYPERNETS_DIR, site, '*', '*', '*', '*', '*.nc')))
    series = []
    for i in range(0, len(files), chunk):
        ds = xr.open_mfdataset(files[i:i + chunk], combine='nested', concat_dim='time').squeeze().load()
        ds['time'] = pd.to_datetime(ds.acquisition_time.values, unit='s')
        series.append(ds)
    hypernets = xr.concat(series, dim='time').sortby('time')
    hypernets = hypernets.rename_dims({'wavelength': 'wl'}).rename_vars({'wavelength': 'wl'})
    hypernets.to_netcdf(ofile)


hypernets_file = opj(HYPERNETS_DIR, SITE, f'{SITE}_hypernets.nc')
if not os.path.exists(hypernets_file):
    build_hypernets_file(SITE, hypernets_file)

hypernets = xr.open_dataset(hypernets_file).load()
hypernets['Rrs_db'] = hypernets[INSITU_VAR] / np.pi      # Rrs = rho_w / pi
hypernets
````

### Normalization to nadir viewing

HYPERNETS observes the water surface with an oblique viewing geometry, while the Sentinel-2 view is
close to nadir. Because of the anisotropy of the water-leaving radiance, the in situ $R_{rs}$ is
converted to nadir viewing with pre-computed factors depending on the solar zenith angle (`sza`) and
wavelength:

$$R_{rs}^{nadir}(\lambda) = f(\theta_s, \lambda)\, R_{rs}^{db}(\lambda)$$

The factors of `DIR_NORM_FILE` were computed for the viewing geometry of the station with a
bio-optical model (see the `model` and `dataset` coordinates of the file).

````python
dir_norm = xr.open_dataarray(DIR_NORM_FILE).interp(wl=hypernets.wl)
hypernets['Rrs_nadir'] = dir_norm.interp(sza=hypernets.solar_zenith_angle) * hypernets.Rrs_db

fig, ax = plt.subplots(figsize=(7, 4))
dir_norm.sel(sza=[20, 40, 60]).plot(hue='sza', ax=ax)
ax.set_ylabel('normalization factor')
ax.set_title(f"{dir_norm.model.values}, {dir_norm.dataset.values}");
````

## Satellite datacube

The FR2030 products are stored as `L2A_DIR/YYYY/MM/DD/<name>.SAFE/<name>.zarr`, with names
`S2_L2A_<time>_<tile>_<processing time>_<version>`. The datacube is cropped to a square box centred on
the station ({meth}`grstbx.SpatioTemp.wktbox`) and cached as a netCDF file. For each image, the
proportion of pixels raised for each flag is added as `flag_<name>` variables (see
{meth}`grstbx.L2grs.get_flag_stats`).

````python
def list_fr2030_products(rootdir, pattern='*.zarr'):
    """
    List the FR2030 products stored as ``rootdir/YYYY/MM/DD/<name>.SAFE/<name>.zarr``.

    The product names are ``S2_L2A_<time>_<tile>_<processing time>_<version>``.

    :return: pandas.DataFrame indexed (and sorted) by acquisition time, with the path of each product
    """
    paths = sorted(glob.glob(opj(rootdir, '*', '*', '*', '*.SAFE', pattern)))
    if not paths:
        raise FileNotFoundError(f'no product found in {rootdir}')
    names = pd.Series([os.path.basename(p).removesuffix('.zarr') for p in paths])
    fields = names.str.split('_', expand=True)
    products = pd.DataFrame({'time': pd.to_datetime(fields[2]), 'tile': fields[3],
                             'processing_time': fields[4],
                             'version': fields.iloc[:, 5:].astype(str).agg('_'.join, axis=1),
                             'path': paths})
    return products.set_index('time').sort_index()
````

````python
products = list_fr2030_products(L2A_DIR)
files = products[START_DATE:END_DATE].path.values
print(f'{len(files)} L2A products from {START_DATE} to {END_DATE}')

clon, clat = hypernets.attrs['site_longitude'], hypernets.attrs['site_latitude']
box = grstbx.SpatioTemp().wktbox(clon, clat, width=BOX_SIZE, height=BOX_SIZE)
bbox = gpd.GeoDataFrame(geometry=gpd.GeoSeries.from_wkt([box]), crs=4326)

os.makedirs(DATACUBE_DIR, exist_ok=True)
dc_path = opj(DATACUBE_DIR, f'{SITE}_FR2030_{START_DATE}_{END_DATE}_L2A_matchup.nc')
if not os.path.exists(dc_path):
    dc = grstbx.L2grs(files)
    dc.get_l2a_datacube(subset=bbox)
    dc.datacube.to_netcdf(dc_path)

dcube = xr.open_dataset(dc_path, decode_coords='all').drop_duplicates('time').load()
dcube
````

````python
fig, axs = plt.subplots(nrows=2, figsize=(15, 7), sharex=True)
for flag, label in [('flag_opac_cirrus', 'opaque cirrus'), ('flag_cloud_p06', 'cloud (P06)'),
                    ('flag_water_swir_visible_index', 'water SWIR-visible index')]:
    axs[0].plot(dcube.time.values, dcube[flag].values, 'o', ms=3, alpha=0.8, label=label)
axs[0].set_ylabel('flagged proportion')
axs[0].legend()
axs[1].plot(dcube.time.values, dcube.mean_solar_zenith_angle.values, 'o', ms=3, c='k')
axs[1].set_ylabel('SZA (deg)')
axs[0].set_title(f'Sentinel-2 images over {SITE}');
````

Statistics of the valid pixels (`mask == 0`) within the box, for each image:

````python
# 'central_wavelength' (actual band centres of S2A / S2B) is not needed here
Rrs = dcube.Rrs.sel(wl=slice(0, WL_MAX)).where(dcube.mask == 0).drop_vars('central_wavelength', errors='ignore')
npix = dcube.sizes['x'] * dcube.sizes['y']
with warnings.catch_warnings():  # images without any valid pixel
    warnings.simplefilter('ignore', RuntimeWarning)
    sat = xr.Dataset({'Rrs_median': Rrs.median(['x', 'y']),
                      'Rrs_q25': Rrs.quantile(0.25, ['x', 'y']).drop_vars('quantile'),
                      'Rrs_q75': Rrs.quantile(0.75, ['x', 'y']).drop_vars('quantile'),
                      'Rrs_mean': Rrs.mean(['x', 'y']),
                      'Rrs_std': Rrs.std(['x', 'y']),
                      'valid_fraction': Rrs.isel(wl=0).count(['x', 'y']) / npix,
                      'sza': dcube.sza.mean(['x', 'y']),
                      'aot550': dcube.aot550.mean(['x', 'y'])})
sat = sat.where(sat.valid_fraction > 0, drop=True)
print(f'{sat.sizes["time"]} images with valid pixels')
````

## Matchups

For each image, the in situ spectra acquired within `MAX_LAG` are summarized by their median,
quartiles and standard deviation, and interpolated to the nominal wavelengths of Sentinel-2.

````python
def insitu_statistics(time):
    window = hypernets.sel(time=slice(time - MAX_LAG, time + MAX_LAG))
    if window.sizes['time'] == 0:
        return None
    stats = {'n_spectra': window.sizes['time']}
    with warnings.catch_warnings():  # wavelengths without valid in situ data
        warnings.simplefilter('ignore', RuntimeWarning)
        for param in ['Rrs_db', 'Rrs_nadir']:
            Rrs = window[param]
            stats.update({f'is{param}_median': Rrs.median('time'),
                          f'is{param}_q25': Rrs.quantile(0.25, 'time').drop_vars('quantile'),
                          f'is{param}_q75': Rrs.quantile(0.75, 'time').drop_vars('quantile'),
                          f'is{param}_std': Rrs.std('time')})
    return xr.Dataset(stats).expand_dims(time=[time])


insitu_stats = [insitu_statistics(time) for time in sat.time.values]
insitu_stats = xr.concat([s for s in insitu_stats if s is not None], dim='time')
insitu_stats = insitu_stats.interp(wl=sat.wl.values)

matchup = xr.merge([sat, insitu_stats], join='inner')
matchup['good'] = matchup.valid_fraction >= MIN_VALID_FRACTION
print(f'{matchup.sizes["time"]} matchups, {int(matchup.good.sum())} with more than '
      f'{MIN_VALID_FRACTION:.0%} of valid pixels')

matchup.to_netcdf(opj(DATACUBE_DIR, f'{SITE}_FR2030_{START_DATE}_{END_DATE}_L2A_matchup_stats.nc'))
matchup
````

## Agreement statistics

````python
def matchup_metrics(x, y):
    """
    Agreement statistics between in situ (``x``) and satellite (``y``) values.

    - slope, intercept and Pearson's r of the linear regression of y vs x,
    - rmse and bias (mean of y - x), in the units of the data,
    - mape: mean absolute percentage error (%),
    - eps: median symmetric accuracy (%) and beta: signed symmetric percentage bias (%)
      (Morley et al., 2018; Eqs. 8-9 in Pahlevan et al., 2020), computed on the
      strictly positive pairs.
    """
    from scipy import stats

    x, y = np.asarray(x, dtype=float), np.asarray(y, dtype=float)
    ok = np.isfinite(x) & np.isfinite(y)
    x, y = x[ok], y[ok]
    regr = stats.linregress(x, y)
    pos = (x > 0) & (y > 0)
    log_ratio = np.log10(y[pos] / x[pos])
    med = np.median(log_ratio)
    return {'N': len(x),
            'slope': regr.slope,
            'intercept': regr.intercept,
            'r': regr.rvalue,
            'rmse': np.sqrt(np.mean((y - x) ** 2)),
            'bias': np.mean(y - x),
            'mape': 100 * np.mean(np.abs(y - x) / np.abs(x)),
            'eps': 100 * (10 ** np.median(np.abs(log_ratio)) - 1),
            'beta': 100 * np.sign(med) * (10 ** np.abs(med) - 1)}


def plot_matchup(ax, x, y, vmin, vmax, fmt='.4f', xerr=None, yerr=None, **kwargs):
    """Scatter plot of y vs x with error bars, 1:1 and regression lines, and the agreement statistics."""
    m = matchup_metrics(x, y)
    if xerr is not None or yerr is not None:
        ax.errorbar(x, y, xerr=xerr, yerr=yerr, fmt='none', color='grey', lw=0.6, capsize=2, zorder=1)
    sc = ax.scatter(x, y, zorder=2, **kwargs)
    ax.set_xlim(vmin, vmax)
    ax.set_ylim(vmin, vmax)
    grstbx.Plotting.set_layout(ax)          # square axes and 1:1 line
    ax.axline((0, m['intercept']), slope=m['slope'], ls='--', lw=1.2, c='grey')
    text = '\n'.join([f"$y = {m['slope']:.2f}x {m['intercept']:+{fmt}}$",
                      f"$r = {m['r']:.3f}$",
                      f"$rmse = {m['rmse']:{fmt}}$",
                      f"$bias = {m['bias']:{fmt}}$",
                      f"$\\epsilon = {m['eps']:.1f}\\%$",
                      f"$\\beta = {m['beta']:+.1f}\\%$",
                      f"$N = {m['N']}$"])
    ax.text(0.97, 0.03, text, fontsize=10, ha='right', va='bottom', transform=ax.transAxes)
    ax.xaxis.set_major_locator(mpl.ticker.MaxNLocator(4))
    ax.yaxis.set_major_locator(mpl.ticker.MaxNLocator(4))
    ax.minorticks_on()
    return sc, m
````

Per band, with the in situ $R_{rs}$ normalized to nadir viewing:

````python
good = matchup.where(matchup.good, drop=True)
vmax = 0.02
cmap = plt.cm.Spectral_r
norm = mpl.colors.Normalize(vmin=440, vmax=870)

wls = good.wl.values
ncols = 3
nrows = int(np.ceil(len(wls) / ncols))
fig, axs = plt.subplots(nrows, ncols, figsize=(ncols * 4.6, nrows * 4.6), squeeze=False)
for ax in axs.flat:
    ax.set_axis_off()
metrics = {}
for ax, wl in zip(axs.flat, wls):
    m = good.sel(wl=wl)
    ax.set_axis_on()
    vmax_band = 1.1 * float(max(m.isRrs_nadir_q75.max(), m.Rrs_q75.max()))
    _, metrics[wl] = plot_matchup(ax, m.isRrs_nadir_median, m.Rrs_median, 0, vmax_band,
                                  xerr=m.isRrs_nadir_std, yerr=m.Rrs_std, color=cmap(norm(wl)), s=18)
    ax.set_title(f'{wl} nm')
    ax.set_xlabel(r'in situ $R_{rs}\ (sr^{-1})$')
    ax.set_ylabel(r'Sentinel-2 $R_{rs}\ (sr^{-1})$')
fig.tight_layout()

pd.DataFrame(metrics).T.round(4)
````

All bands, before and after the normalization of the in situ data to nadir viewing:

````python
df = good[['Rrs_median', 'Rrs_std', 'isRrs_db_median', 'isRrs_nadir_median']].to_dataframe().dropna().reset_index()

fig, axs = plt.subplots(ncols=2, figsize=(14, 5.5), layout='constrained')
for ax, insitu_var, title in zip(axs, ['isRrs_db_median', 'isRrs_nadir_median'],
                                 ['in situ from the data base', 'in situ normalized to nadir']):
    sc, _ = plot_matchup(ax, df[insitu_var], df.Rrs_median, -0.0005, vmax, yerr=df.Rrs_std,
                         c=df.wl, cmap=cmap, norm=norm, s=10)
    ax.set_xlabel(r'in situ $R_{rs}\ (sr^{-1})$')
    ax.set_ylabel(r'Sentinel-2 $R_{rs}\ (sr^{-1})$')
    ax.set_title(title)
fig.colorbar(sc, ax=axs, shrink=0.8, label='wavelength (nm)');
````

## Comparison with the GRS products

Median $R_{rs}$ within the box of the FR2030 products against the GRS products of the HYPERNETS
tutorial (GRS v2.1.7, `GRS_STATS`), for the images common to both data sets.

````python
grs = xr.open_dataset(GRS_STATS)
common = np.intersect1d(matchup.time.values.astype('datetime64[m]'), grs.time.values.astype('datetime64[m]'))
fr = matchup.assign_coords(time=matchup.time.values.astype('datetime64[m]')).sel(time=common)
ref = grs.assign_coords(time=grs.time.values.astype('datetime64[m]')).sel(time=common)
print(f'{len(common)} images in common')

df = xr.Dataset({'fr2030': fr.Rrs_median, 'grs': ref.Rrs_median}).to_dataframe().dropna().reset_index()
fig, ax = plt.subplots(figsize=(7.5, 6), layout='constrained')
sc, _ = plot_matchup(ax, df.grs, df.fr2030, -0.0005, 1.1 * float(df[['grs', 'fr2030']].max().max()),
                     c=df.wl, cmap=cmap, norm=norm, s=14)
ax.set_xlabel(r'GRS $R_{rs}\ (sr^{-1})$')
ax.set_ylabel(r'FR2030 $R_{rs}\ (sr^{-1})$')
ax.set_title('FR2030 vs GRS, median within the box')
fig.colorbar(sc, ax=ax, shrink=0.8, label='wavelength (nm)');
````

**Notes**

- Few FR2030 products are available on this site (March 2021): the statistics are indicative only.

- The in situ spectra are linearly interpolated to the nominal Sentinel-2 wavelengths; a
  convolution with the spectral response functions of each band (S2A and S2B) would be more
  accurate, especially for the red-edge bands.
- The HYPERNETS `quality_flag` is not used here.
- The `eps` and `beta` metrics are less sensitive to outliers than `rmse` and `bias`; they are
  computed in log space and expressed in %.

**References**

  Sentinel-3 (OLCI) in inland and coastal waters: a machine-learning approach. *Remote Sensing of
  Environment*, 240, 111604.
- Ruddick, K. G., et al. (2006). Seaborne measurements of near infrared water-leaving reflectance:
